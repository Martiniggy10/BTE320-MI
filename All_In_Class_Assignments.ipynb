Recursive function: string reversal

```python
def reverse_string(s):
  if len(s) <= 1:
    return s
  return reverse_string(s[1:]) + s[0]
string = input("Enter a string: ")
print(reverse_string(string))
```

Greeting Class

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last

  def display(self, formal=False):
    if formal:
      print(f'Hello {self.first} {self.last}')
    else:
      print(f'Hello {self.first}')

def __str__(self):
  return str(self.first) + "," + str(self.last)

greeting = Greeting('John', 'Smith')
greeting.display(formal=True)
```

```python
class Student:
  def __init__(self, name, number):
    self.name = name
    self.number = number
    self.courses = []

  def enroll(self,course):
    if course not in self.courses:
      self.courses.append(course)
    else:
      print(f'You have already enrolled in {course}')

  def get_courses(self):
    return self.courses

s1= Student('Bob',100)
s2= Student('Alice',101)
s1.name, s2.name
```

Class Point

```python
import math

class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def distance(self, other):
    dx = (self.x - other.x) ** 2
    dy = (self.y - other.y) ** 2
    dist = math.sqrt(dx + dy)
    return dist

p1 = Point(1,7)
p2 = Point(4,3)
p1.distance(p2)
```

Class Rectangle

```python
class Rectangle:
  def __init__(self):
    self.c1 = Point(0, 0)
    self.c2 = Point(0, 2)
    self.c3 = Point(2, 0)
    self.c4 = Point(2, 2)
    self.width = self.calculate_width()
    self.length = self.calculate_length()

  def calculate_width(self):
    return self.c1.distance (self.c3)

  def calculate_length(self):
    return self.c1.distance (self.c2)

  def area(self):
    return f'Area = {self.width * self.length} sq ft.'


square = Rectangle()
square.area()
```
